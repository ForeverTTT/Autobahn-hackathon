# ChatSession 多轮对话测试

测试 AlpineFlow AI Agent 的多轮对话功能。

## 0. 环境设置

```python
import sys
from pathlib import Path

ROOT = Path.cwd()
if ROOT.name == "agent":
    ROOT = ROOT.parent
if str(ROOT) not in sys.path:
    sys.path.insert(0, str(ROOT))

from agent import ChatSession

print("环境设置完成!")
```

## 1. 基础对话 + 追问原因

```python
# 创建会话
session = ChatSession(verbose=True)

# 第一轮: 新查询
print("=" * 50)
print("用户: 周六去萨尔茨堡")
print("=" * 50)
print(session.chat("周六去萨尔茨堡"))
```

```python
# 第二轮: 追问原因
print("=" * 50)
print("用户: 为什么")
print("=" * 50)
print(session.chat("为什么"))
```

```python
# 第三轮: 询问细节
print("=" * 50)
print("用户: 那天天气怎么样")
print("=" * 50)
print(session.chat("那天天气怎么样"))
```

## 2. 修改计划

```python
# 新会话
session2 = ChatSession(verbose=True)

print("=" * 50)
print("用户: 周六去萨尔茨堡")
print("=" * 50)
print(session2.chat("周六去萨尔茨堡"))
```

```python
# 修改计划
print("=" * 50)
print("用户: 改成周日呢")
print("=" * 50)
print(session2.chat("改成周日呢"))
```

## 3. 假设性问题

```python
# 新会话
session3 = ChatSession(verbose=True)

print("=" * 50)
print("用户: 明天去萨尔茨堡")
print("=" * 50)
print(session3.chat("明天去萨尔茨堡"))
```

```python
# 假设问题
print("=" * 50)
print("用户: 如果我只能下午2点出发呢")
print("=" * 50)
print(session3.chat("如果我只能下午2点出发呢"))
```

## 4. 询问返程

```python
# 新会话
session4 = ChatSession(verbose=True)

print("=" * 50)
print("用户: 周末带家人去萨尔茨堡玩")
print("=" * 50)
print(session4.chat("周末带家人去萨尔茨堡玩"))
```

```python
# 询问返程
print("=" * 50)
print("用户: 下午回来堵不堵")
print("=" * 50)
print(session4.chat("下午回来堵不堵"))
```

## 5. 全局会话 (简便方式)

```python
from agent import chat_session, clear_session

# 清除之前的会话
clear_session()

# 使用全局会话
print("=" * 50)
print("用户: 后天去因斯布鲁克")
print("=" * 50)
print(chat_session("后天去因斯布鲁克"))
```

```python
# 追问 (自动关联上一轮)
print("=" * 50)
print("用户: 为什么推荐这个时间")
print("=" * 50)
print(chat_session("为什么推荐这个时间"))
```

## 6. 查看会话数据

```python
from agent import get_session

session = get_session()

# 查看当前计划
print("当前计划:")
print(session.plan[:500] if session.plan else "无")

print()
print("对话历史:")
for msg in session.history:
    print(f"  [{msg.role}]: {msg.content[:50]}...")
```
